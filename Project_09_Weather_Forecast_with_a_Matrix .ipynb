## The problem

Tomorrow's weather depends only on today's weather. We write the chances in a **matrix** `P`. Row `i` gives the chances for tomorrow if today is state `i`.

Example with 2 states (Sunny, Rainy):

```python
P = [[0.8, 0.2],     # today Sunny: 80% Sunny tomorrow, 20% Rainy
     [0.4, 0.6]]     # today Rainy: 40% Sunny tomorrow, 60% Rainy
```

A **distribution** is a list of chances, like `[1, 0]` (today is Sunny for sure). To find tomorrow's distribution:

$$\text{new}[j] = \text{dist}[0] \cdot P[0][j] + \text{dist}[1] \cdot P[1][j] + \dots$$

**Example:** `dist = [1, 0]`.

- `new[0] = 1 * 0.8 + 0 * 0.4 = 0.8`
- `new[1] = 1 * 0.2 + 0 * 0.6 = 0.2`

So tomorrow is `[0.8, 0.2]`. The day after is `[0.72, 0.28]`.

After many days, the numbers stop changing. This is the **long-run** distribution.

## Your code

````python
P = [[0.8, 0.2], [0.4, 0.6]]

def next_day(dist, P):
    n=len(dist)
    new = [0] * n
    for j in range(n):
      for i in range(n):
        new[j] = round(new[j]+ dist[i] * P[i][j],4)
    return new


def after_days(start, P, days):
    dist = start
    for jour in range(days):
      dist = next_day(dist, P)
    return dist


# Tasks 3 and 4: write your code here
#new matrix The states are Sunny, Cloudy, Rainy.
P3 = [[0.7, 0.2, 0.1], [0.3, 0.4, 0.3], [0.2, 0.3, 0.5]]
for d in [1,2,5,10,50]:
  res=after_days([1,0,0],P3,d)
  resultat = []
  for x in res:
    resultat.append(round(x, 4))
  print(d, resultat)
````

````python
#methode 2 avec les listes comprehension
def next_day(dist, P):
    # i compute the new dist
    return [sum(round(dist[j]*P[j][i], 4)  for j in range(len(dist))) for i in range(len(P[0]))]

def after_days(start, P, days):
    new_dist=start
    for k in range(days):
        new_dist=next_day(new_dist, P)
    return new_dist
````
